```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

### normal distribution

```python
mu = 0
sigma = 0.1

data_n = np.random.normal(mu,sigma,size = 1000)
```

```python
plt.hist(data_n,30)
plt.show()
```

```python
sns.distplot(data_n)
plt.show()
```

### unifrom distribution

```python
data_u = np.random.uniform(mu,sigma,size = 1000)
```

```python
plt.hist(data_u)
plt.show()
```

```python
sns.distplot(data_u)
plt.show()
```

```python
### log normal distribution
```

```python
data_l = np.random.lognormal(mean = 0,sigma = 0.5,size = 1000)
data_l[np.where(abs(data_l)< 5)]
plt.hist(data_l,bins = 100)
plt.show()
```

```python
sns.boxplot(data_l) # since the log normal is right skewed so the data have outliers present
```

```python
sns.distplot(data_l)
```

```python
#  log normal distribution
sns.distplot(np.log(data_l))
```

### chi square distribution

```python
data_c = np.random.chisquare(df = 30,size = 10000)
data_c = data_c[np.where(abs(data_c)<100)]
plt.hist(data_c, bins = 1000)
plt.show()
```

### t distribution

```python
data_t = np.random.standard_t(df = 5,size = 10000)
data_t = data_t[np.where(abs(data_t)< 100)]
```

```python
plt.hist(data_t,bins = 1000)
plt.show()
```

### f distribution

```python
data_f = np.random.f(dfnum = 100,dfden = 100 , size = 1000)
```

```python
data_f = data_f[np.where(abs(data_f)<10)]
```

```python
plt.hist(data_f,bins= 100)
plt.show()
```
